```python
import os
import requests
import zipfile
from io import BytesIO

symbol = "BTCUSDT"
period = "2024-01"
url = f"https://data.binance.vision/data/spot/monthly/trades/{symbol}/{symbol}-trades-{period}.zip"

bronze_path = os.path.join("data_lake", "bronze", symbol)
os.makedirs(bronze_path, exist_ok=True)

print(f"Starting download for {symbol} | Period: {period}...")

try:
    response = requests.get(url, stream=True, timeout=90)
    
    if response.status_code == 200:
        print("Download complete! Extracting zip contents...")
        
        with zipfile.ZipFile(BytesIO(response.content)) as z:
            z.extractall(bronze_path)
            
        print(f"Successfully extracted files to: {bronze_path}")
        
        saved_files = os.path.listdir(bronze_path)
        print(f"Verified files in Bronze layer: {saved_files}")
        
    else:
        print(f"Failed to download. HTTP Status Code: {response.status_code}")
        
except Exception as e:
    print(f"An error occurred during ingestion: {str(e)}")
```
