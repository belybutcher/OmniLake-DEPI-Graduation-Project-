```python
from pyspark.sql import SparkSession

spark = SparkSession.builder.appName("Verify-Gold").getOrCreate()

for name, path in [
    ("fct_ohlcv_1h", "data_lake/gold/fct_ohlcv_1h/BTCUSDT"),
    ("fct_ohlcv_1h_unified", "data_lake/gold/fct_ohlcv_1h_unified"),
    ("fct_trades_daily", "data_lake/gold/fct_trades_daily/BTCUSDT"),
]:
    df = spark.read.parquet(path)
    print(f"=== {name} (count={df.count()}) ===")
    df.printSchema()
    df.show(5, truncate=False)

spark.stop()
```
